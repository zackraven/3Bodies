```python
import time
import numpy as np
import matplotlib.pyplot as plt
x1, y1 = [], []
x2, y2 = [], []
x3, y3 = [], []
h = 0.05
G =  10
Mass1 = 10
Mass2 = 10
Mass3 = 10
r1i = np.array([10, 0])
r2i = np.array([-10, 0])
r3i = np.array([0, 0])
v1i = np.array([1, -2])
v2i = np.array([-1, 2])
v3i = np.array([1, 0])
Ac1i = ((G*Mass2)*(r2i-r1i)/np.linalg.norm(r2i - r1i)**3)+((G*Mass3)*(r3i-r1i)/np.linalg.norm(r3i - r1i)**3)
Ac2i = ((G*Mass1)*(r1i-r2i)/np.linalg.norm(r1i - r2i)**3)+((G*Mass3)*(r3i-r2i)/np.linalg.norm(r3i - r2i)**3)
Ac3i = ((G*Mass1)*(r1i-r3i)/np.linalg.norm(r1i - r3i)**3)+((G*Mass2)*(r2i-r3i)/np.linalg.norm(r2i - r3i)**3)
i=0

for i in range(150):
    r1n = r1i + v1i*h+0.5*Ac1i*h**2
    r2n = r2i + v2i*h+0.5*Ac2i*h**2
    r3n = r3i + v3i*h+0.5*Ac3i*h**2
    Ac1n = ((G*Mass2)*(r2n-r1n)/np.linalg.norm(r2n - r1n)**3)+((G*Mass3)*(r3n-r1n)/np.linalg.norm(r3n - r1n)**3)
    Ac2n = ((G*Mass1)*(r1n-r2n)/np.linalg.norm(r1n - r2n)**3)+((G*Mass3)*(r3n-r2n)/np.linalg.norm(r3n - r2n)**3)
    Ac3n = ((G*Mass1)*(r1n-r3n)/np.linalg.norm(r1n - r3n)**3)+((G*Mass2)*(r2n-r3n)/np.linalg.norm(r2n - r3n)**3)
    v1n = v1i + 0.5*h*(Ac1i +Ac1n)
    v2n = v2i + 0.5*h*(Ac2i +Ac2n)
    v3n = v3i + 0.5*h*(Ac3i +Ac3n)
    print ("posistion 1:",r1n)
    print ("posistion 2:",r2n)
    print ("posistion 3:",r3n)
    print ("velocity 1:",v1n)
    print ("velocity 2:",v2n)
    print ("velocity 3:",v3n)
    r1i = r1n
    r2i = r2n
    r3i = r3n
    v1i = v1n
    v2i = v2n
    v3i = v3n
    Ac1i = Ac1n
    Ac2i =Ac2n
    Ac3i =Ac3n
    x1.append(r1n[0])
    y1.append(r1n[1])
    x2.append(r2n[0])
    y2.append(r2n[1])
    x3.append(r3n[0])
    y3.append(r3n[1])
    i= i+1
plt.figure(figsize=(6,6))

plt.plot(x1, y1, 'b-', label='Particle 1')
plt.plot(x2, y2, 'r-', label='Particle 2')
plt.plot(x3, y3, 'g-', label='Particle 3')

plt.scatter(x1[0], y1[0], c='b', marker='o')
plt.scatter(x2[0], y2[0], c='r', marker='o')
plt.scatter(x3[0], y3[0], c='g', marker='o')

plt.scatter(x1[-1], y1[-1], c='b', marker='x', s=100)
plt.scatter(x2[-1], y2[-1], c='r', marker='x', s=100)
plt.scatter(x3[-1], y3[-1], c='g', marker='x', s=100)

plt.axis('equal')
plt.grid(True)
plt.legend()
plt.show()

```
